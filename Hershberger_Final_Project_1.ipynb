**Research Question**

This research project examines the effect that setting a minimum wage has on a state, specifically the unemployment and poverty rates. This study will explore whether higher and lower minimum wages have an overall positive or negative effect on a state's economy. Specifically, do they improve the state's economy by reducing poverty or have a negative effect by increasing unemployment?

Opponents of raising the minimum wage argue that the increased labor market costs force businesses to reduce hiring or lay off workers, which could lead to higher unemployment. They also say that increasing the minimum wage has little effect on reducing poverty because businesses would have to reduce hiring. Supporters on the other hand, argue that higher minimum wages help lift workers out of poverty by increasing earnings for low-income households. They also argue that higher wages increase consumer spending, which can stimulate economic growth and, as a result of that increase employment.



My research question is: Do states with higher minimum wages experience lower poverty rates but higher unemployment rates?




**Data Cleaning **

```python
import pandas as pd
import seaborn as sns
```

This data shows the minimum wage for each state in the year 2026. It is collected from the Federal Reserve Economic Data website.

```python
url = "https://fred.stlouisfed.org/release/tables?eid=243906&rid=387"

tables = pd.read_html(url)

mw = tables[0]
```

```python
mw.columns = [
    "id",
    "State",
    "Period",
    "minimum_wage",
    "previous_period",
    "year_ago"
]
```

```python
mw_2026 = mw[mw["Period"] == "2026"].copy()

mw_2026 = mw_2026[["State", "minimum_wage"]]

mw_2026["minimum_wage"] = pd.to_numeric(
    mw_2026["minimum_wage"]
)
```

```python
len(mw_2026)
```

```python
missing_states = pd.DataFrame({
    "State": [
        "Alabama",
        "Georgia",
        "Louisiana",
        "Mississippi",
        "South Carolina",
        "Tennessee",
        "Wyoming"
    ],
    "minimum_wage": [7.25] * 7
})

mw_2026 = pd.concat(
    [mw_2026, missing_states],
    ignore_index=True
)
```

```python
mw_2026 = mw_2026[
    ~mw_2026["State"].isin([
        "Federal Minimum Wage Rate",
        "District of Columbia",
        "Guam",
        "Puerto Rico",
        "U.S. Virgin Islands"
    ])
]
```

```python
len(mw_2026)

```

```python
mw_2026.head(15)
```

The first step of this process was to scrape the minimum wage table directly from the website. Because the webpage contained multiple tables, the first table was selected and stored as a pandas data frame for further cleaning. The variables were then renamed to make the dataset easier to understand. The new variable names identified the state, year, minimum wage, previous value, and year-ago value. After renaming the variables, I decided that I only needed observations from 2026, since this study focuses on the current minimum wage policies for each state. So I cleaned the dataset further to only include the state name and the minimum wage for that specific state. After that, I converted minimum wage values to numeric variables.

After completing these steps and running a quick check, I noticed that there were several states that did not report a state minimum wage because they followed the federal minimum wage of $7.25 per hour. To fix this, I manually added these states with their minimum wages. Finally, the dataset included several observations that were not U.S. states, including the District of Columbia and several U.S. territories. Because I wanted to solely focus on just the 50 states, I removed them from the dataset. This then produced a complete state-level minimum wage dataset.


This data was collected through the World Population Review website.

---



```python
url = "https://worldpopulationreview.com/state-rankings/poverty-rate-by-state"

tables = pd.read_html(url)

poverty = tables[0]
```

```python
poverty = poverty[
    ["State", "% Of Population Below Poverty Level↓"]
]
```

```python
poverty.columns = [
    "State",
    "poverty_rate"
]
```

```python
poverty["poverty_rate"] = (
    poverty["poverty_rate"]
    .str.replace("%", "", regex=False)
)
```

```python
poverty["poverty_rate"] = pd.to_numeric(
    poverty["poverty_rate"]
)
```

```python
poverty = poverty[
    ~poverty["State"].isin([
        "District of Columbia",
        "Puerto Rico"
    ])
]
```

```python
poverty = poverty[
    poverty["State"] != "United States"
]
```

```python
len(poverty)
```

Like the minimum wage data, the first step I did in this process was to scrape the poverty table directly from the webpage and into Python. The first table on the webpage contained state-level poverty statistics and was stored in the data frame named poverty. The original dataset contained several variables that were not needed for this study, so I removed all of them with the exception of state name and the percentage of the population that were living below the poverty line. The poverty rate variables also had long and difficult-to-use column names. Therefore, those variables were renamed to State and poverty_rate to improve readability and make the dataset easier to work with. The poverty rate values were originally stored as text because they included percentage symbols. The percentage signs were removed from each observation, and the variable was converted into a numeric data type.


This data was colllected through the USA symbol website, which provides recent state-level unemployment numbers.

```python
url = "https://usasymbol.com/rankings/economy/states-by-unemployment"

tables = pd.read_html(url)

unemployment = tables[0]
```

```python
unemployment = unemployment[
    ["State", "Unempl. Rate"]
]
```

```python
unemployment.columns = [
    "State",
    "unemployment_rate"
]
```

```python
unemployment["unemployment_rate"] = (
    unemployment["unemployment_rate"]
    .astype(str)
    .str.replace("%", "", regex=False)
)
```

```python
unemployment["unemployment_rate"] = pd.to_numeric(
    unemployment["unemployment_rate"]
)
```

```python
len(unemployment)
```

Similar to the other two datasets, the first step in this process was to scrape the unemployment table directly from the webpage and import it into Python. The table contained state-level unemployment statistics and was stored in the data frame unemployment. The original table also contained information that was not needed, and because of that I removed all the unnecessary information, only leaving the state name and unemployment rate variables.
The unemployment rate variable was renamed to unemployment_rate to create a shorter variable name. The unemployment rates were originally stored as text because the values included percentage symbols. Therefore, the percentage signs were removed, and the variable was converted to a numeric data type. I also kept the state variable, as I knew that I was going to need it later on to merge all three datasets.




```python
df = mw_2026.merge(
    poverty,
    on="State"
)

df = df.merge(
    unemployment,
    on="State"
)

len(df)
```

Once I finished cleaning all the three datasets, I merged them using the state name as the common identifier. Combining the datasets created a single state-level dataset that contained all the variables that are needed for the analysis. The final dataset includes observations for all 50 states including the minimum wage for that state, the poverty rate of that state, and the unemployment rate of the state.
